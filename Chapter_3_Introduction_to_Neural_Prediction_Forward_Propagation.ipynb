## 1. Step 1: Predict

This chapter focuses on the prediction step in a neural network. A neural network receives input data and processes it to produce a prediction.

The input data must contain enough information for the network to make a meaningful prediction. The structure of the network depends on the shape of the input and output data.

The basic prediction process consists of input data, a machine or neural network, and an output prediction.

## 2. A Simple Neural Network Making a Prediction

A simple neural network can make a prediction by multiplying an input value by a weight.

The weight determines how strongly the input affects the prediction. In this example, the network uses the average number of toes per baseball player to predict a result.

```python
weight = 0.1

def neural_network(input, weight):
    prediction = input * weight
    return prediction

number_of_toes = [8.5, 9.5, 10, 9]

input = number_of_toes[0]
pred = neural_network(input, weight)

print(pred)
```

## 3. What Is a Neural Network?

A neural network is a system that uses weights to process input data and make predictions.

In the simplest example, a single weight is multiplied by an input value. More complex neural networks can contain many weights and combine multiple inputs.

The input represents information collected from the real world, while the prediction is the output generated by the network.

## 4. What Does This Neural Network Do?

The network multiplies the input by its weight to generate a prediction.

For example, when the input is 8.5 and the weight is 0.1, the prediction is 0.85.

At this stage, the network does not learn automatically. It uses the weight that has already been provided to calculate the prediction.

## 5. Making a Prediction with Multiple Inputs

A neural network can use multiple inputs to make a single prediction.

Each input has its own corresponding weight. The network multiplies each input by its weight and adds the results together.

This process is called a weighted sum. It allows the network to combine information from different input features.

## 6. Multiple Inputs: What Does This Neural Network Do?

The network calculates a weighted sum of the input values.

Each input contributes to the final prediction according to its value and corresponding weight. An input with a weight of zero does not contribute to the result.

The order of the inputs must match the order of the weights. Changing their positions can change the prediction.

```python
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output

weights = [0.1, 0.2, 0]

def neural_network(input, weights):
    pred = w_sum(input, weights)
    return pred

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes[0], wlrec[0], nfans[0]]
pred = neural_network(input, weights)

print(pred)
```

```python
import numpy as np

weights = np.array([0.1, 0.2, 0])

def neural_network(input, weights):
    pred = input.dot(weights)
    return pred

toes = np.array([8.5, 9.5, 9.9, 9.0])
wlrec = np.array([0.65, 0.8, 0.8, 0.9])
nfans = np.array([1.2, 1.3, 0.5, 1.0])

input = np.array([toes[0], wlrec[0], nfans[0]])
pred = neural_network(input, weights)

print(pred)
```

## 8. Predicting with Multiple Inputs and Outputs

A neural network can combine multiple inputs and generate multiple outputs.

Each input is connected to each output through a corresponding weight. The network calculates weighted sums to generate its predictions.

This structure allows the network to use several input features to predict different outcomes simultaneously.

## 9. Multiple Inputs and Outputs: How Does It Work?

The network uses a matrix of weights to connect multiple inputs to multiple outputs.

Each output receives contributions from the input values through their corresponding weights. The weighted contributions are combined to generate the output predictions.

The arrangement of the weights determines how the input information contributes to each output.

## 10. Predicting on Predictions

Neural networks can be stacked so that the output of one network becomes the input of another network.

The first network processes the original input and produces an intermediate prediction. The next network uses that prediction to calculate another output.

This approach allows neural networks to build more complex prediction systems from simpler components.

## 11. A Quick Primer on NumPy

NumPy is a Python library designed for efficient numerical calculations. It provides arrays and mathematical operations that simplify computations involving multiple values.

In this chapter, NumPy's dot product operation is used to calculate a weighted sum without writing a separate loop.

NumPy becomes especially useful when neural networks contain many inputs, outputs, and weights.
