## 1. What Is Deep Learning?

Deep learning is a subset of machine learning that primarily uses artificial neural networks. These networks are loosely inspired by the human brain and are used to solve various practical problems.

Deep learning is applied in several fields, including:

- Computer vision, which processes and analyzes images.
- Natural language processing, which works with human language and text.
- Automatic speech recognition, which converts spoken language into text.

Deep learning is part of machine learning, which belongs to the broader field of artificial intelligence. Many deep learning applications focus on solving specific real-world problems.

## 2. What Is Machine Learning?

Machine learning is a field of computer science that enables computers to learn patterns from data without being explicitly programmed for every task.

Machines observe patterns in data and use those patterns to perform tasks or make predictions.

There are two main types of machine learning discussed in this chapter:

- Supervised machine learning.
- Unsupervised machine learning.

Both approaches allow machines to learn from data, but they differ in the type of information available during the learning process.

## 3. Supervised Machine Learning

Supervised machine learning learns the relationship between two datasets. One dataset serves as the input, while the other serves as the output that the algorithm attempts to predict.

For example, a model can learn the relationship between Monday stock prices and Tuesday stock prices. After learning from historical data, the model can use Monday's prices to predict Tuesday's prices.

The main purpose of supervised learning is to transform known input data into useful output predictions.

Supervised learning is widely used for classification and prediction tasks.

## 4. Unsupervised Machine Learning

Unsupervised machine learning identifies patterns or structures in data without being provided with predefined correct answers.

One example is clustering, which groups data points according to similarities.

For example, a dataset containing different objects can be separated into groups based on their characteristics. The algorithm identifies the groups, while the meaning of each group must be interpreted by the person analyzing the results.

Unlike supervised learning, unsupervised learning does not attempt to reproduce a known output. Instead, it discovers patterns within the available data.

## 5. Parametric vs. Nonparametric Learning

Machine learning algorithms can also be classified as parametric or nonparametric, depending on how they represent learned patterns.

### Parametric Models

A parametric model has a fixed number of parameters. These parameters represent the patterns the model learns from data.

### Nonparametric Models

A nonparametric model allows its number of parameters to depend on the dataset.

The main difference is how the learned patterns are represented. Parametric models use a fixed number of parameters, while nonparametric models can adapt their representation according to the data.

These categories can be combined with supervised and unsupervised learning.

## 6. Supervised Parametric Learning

Supervised parametric learning uses a fixed number of parameters to learn the relationship between input data and output predictions.

The learning process can be imagined as adjusting a set of knobs. Each knob represents a parameter that affects the model's predictions.

During training, the parameter values are adjusted to improve the relationship between the input data and the expected output.

This approach combines the characteristics of supervised learning, which learns input-output relationships, and parametric learning, which uses a fixed number of parameters.

## 7. Unsupervised Parametric Learning

Unsupervised parametric learning identifies patterns in data without predefined output labels while using a fixed set of parameters to represent those patterns.

The chapter introduces counting and probability as ways to represent patterns in data.

Counts can be used as parameters, while normalized counts can represent probabilities.

This approach allows a model to describe patterns in data without requiring predefined correct answers for each input.

## 8. Nonparametric Learning

Nonparametric learning allows the number of parameters used to represent patterns to depend on the dataset.

Unlike parametric models, which use a fixed number of parameters, nonparametric models can adjust their representation according to the data they observe.

The chapter uses counting as an intuitive way to understand nonparametric learning.

The important concept is that the model's representation is not restricted to a fixed number of parameters in the same way as a parametric model.
