
## Introduction to Automatic Gradient Computation (Autograd)

Gradient computation is an important part of training neural networks.

A gradient indicates how a function's output changes when its input changes. Backpropagation uses gradients to determine how model parameters should be adjusted to reduce prediction error.

Automatic differentiation tracks mathematical operations and computes derivatives systematically, reducing the need to calculate every derivative manually.


```python

# Function: y = x^2
x = 3.0

y = x ** 2
gradient = 2 * x

print("Input x:", x)
print("Output y:", y)
print("Gradient dy/dx:", gradient)
```


## A Quick Checkpoint

Before building an automatic differentiation system, it is useful to understand how derivatives flow through mathematical operations.

For a function composed of multiple operations, the chain rule combines the derivatives of the individual operations.

A deep learning framework can use this principle to calculate gradients automatically.



## Tensors That Are Used Multiple Times

A tensor may contribute to multiple operations in a computational graph.

When this happens, the gradient of the final result with respect to that tensor must account for every path through which the tensor affects the output.

This is important when implementing automatic differentiation because a tensor may be reused in different parts of a neural network.



## Upgrading Autograd to Support Multiuse Tensors

An automatic differentiation system must accumulate gradients from multiple computational paths.

If a variable contributes to more than one operation, its total gradient is obtained by adding the gradient contributions from those operations.

Gradient accumulation enables the framework to handle computational graphs in which tensors are reused.



## How Does Addition Backpropagation Work?

For an addition operation, the gradient from the output is passed to both input variables.

If z = x + y, then the derivative of z with respect to x is 1, and the derivative of z with respect to y is also 1.

Using the chain rule, the incoming gradient is therefore propagated to both inputs.


```python

x = 2.0
y = 4.0

z = x + y

dz_dx = 1.0
dz_dy = 1.0

print("z =", z)
print("dz/dx =", dz_dx)
print("dz/dy =", dz_dy)
```


## Adding Support for Negation

Negation changes the sign of a value.

For the operation z = -x, the derivative of z with respect to x is -1.

During backpropagation, the incoming gradient is multiplied by -1 before being passed to the input.

This operation is one of the building blocks needed to support more complex mathematical expressions.



## Adding Support for Additional Functions

A deep learning framework needs to support multiple mathematical operations, each with its own derivative rule.

By combining these operations and applying the chain rule, a framework can compute gradients for more complicated functions.

These capabilities form the foundation for implementing automatic differentiation and training neural networks.



## Using Autograd to Train a Neural Network

A neural network learns by adjusting its parameters based on the prediction error.

Automatic differentiation computes the gradients of the loss with respect to the parameters. An optimizer then uses those gradients to update the parameters.

This chapter builds toward a framework that can perform these steps automatically.



## Adding Automatic Optimization

An optimization algorithm updates model parameters using the gradients calculated during backpropagation.

A basic approach is gradient descent, which adjusts a parameter in the opposite direction of its gradient.

The learning rate controls the size of each update. Choosing an appropriate learning rate is important for stable and effective training.



## Adding Support for Layer Types

Neural networks are commonly organized into layers.

Each layer performs a specific computation, such as a linear transformation, an activation function, or an embedding lookup.

Supporting different layer types makes a framework more modular and easier to extend.



## Layers That Contain Layers

A framework can organize multiple layers into a larger model. This allows complex neural networks to be constructed from reusable components.

## Loss-Function Layers

A loss function measures the difference between predictions and target values.

The loss provides the objective that the optimization process attempts to minimize. Its gradient is propagated backward through the network to update the parameters.



## Nonlinearity Layers

Nonlinear activation functions allow neural networks to model relationships that cannot be represented by a purely linear transformation.

## The Embedding Layer

An embedding layer maps discrete indices to trainable vectors. It is useful for representing words or other categorical inputs in a numerical form.

## The Recurrent Neural Network Layer

A recurrent layer processes sequential inputs while carrying a hidden state across time steps. It reuses parameters across sequence positions.
