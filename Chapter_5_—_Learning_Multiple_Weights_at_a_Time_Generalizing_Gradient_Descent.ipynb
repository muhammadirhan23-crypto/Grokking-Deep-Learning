## Chapter Overview

This chapter explains how gradient descent can update multiple weights in a neural network. It covers networks with multiple inputs, multiple outputs, and combinations of both. It also explores how weight values influence predictions and how weighted sums are calculated.

## 1. Gradient Descent Learning with Multiple Inputs

A neural network can use multiple input features to make one prediction. Each input has its own weight, and the weighted inputs are combined to produce the prediction.

During learning, the network calculates the prediction error and updates each weight. The same gradient descent principle used for a single weight can be applied to several weights.

```python
def w_sum(a, b):
    assert len(a) == len(b)
    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output


def neural_network(input, weights):
    return w_sum(input, weights)


weights = [0.1, 0.2, -0.1]
input = [8.5, 0.65, 1.2]

prediction = neural_network(input, weights)
print(prediction)
```

## 2. Gradient Descent with Multiple Inputs Explained

Each weight contributes to the final prediction according to its corresponding input. To update the weights, the network calculates the prediction error, determines the direction of correction, and calculates a weight update for each input.

The weight updates depend on both the shared error signal and the individual input values. Therefore, different weights can change by different amounts.

```python
weights = [0.1, 0.2, -0.1]
input = [8.5, 0.65, 1.2]
true = 1
alpha = 0.01

pred = neural_network(input, weights)

error = (pred - true) ** 2
delta = pred - true

weight_deltas = [x * delta for x in input]

for i in range(len(weights)):
    weights[i] -= alpha * weight_deltas[i]

print("Prediction:", pred)
print("Error:", error)
print("Updated weights:", weights)
```

## 3. Let’s Watch Several Steps of Learning

Gradient descent repeatedly predicts, calculates the error, and updates the weights. As the process continues, the weights can move toward values that reduce the prediction error.

The learning rate controls the size of each update. A suitable learning rate helps the model learn without making excessively large corrections.

```python
weights = [0.1, 0.2, -0.1]
input = [8.5, 0.65, 1.2]
true = 1
alpha = 0.01

for iteration in range(10):
    pred = neural_network(input, weights)
    error = (pred - true) ** 2
    delta = pred - true

    weight_deltas = [x * delta for x in input]

    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]

    print(
        "Iteration:", iteration + 1,
        "Error:", error,
        "Weights:", weights
    )
```

## 4. Freezing One Weight: What Does It Do?

Freezing a weight means preventing that weight from changing during training. Other weights can continue to be updated.

This experiment helps illustrate how individual weights affect the error. The network’s goal is to find a combination of weights that produces a low error for the training data.

```python
weights = [0.1, 0.2, -0.1]
input = [8.5, 0.65, 1.2]
true = 1
alpha = 0.01

for iteration in range(10):
    pred = neural_network(input, weights)
    delta = pred - true

    weight_deltas = [x * delta for x in input]

    # Keep the first weight fixed.
    for i in range(1, len(weights)):
        weights[i] -= alpha * weight_deltas[i]

    print("Iteration:", iteration + 1, "Weights:", weights)
```

## 5. Gradient Descent Learning with Multiple Outputs

A neural network can produce several predictions from a single input. Each output has its own weight and prediction error.

The network calculates the error for each output and updates the corresponding weights. This extends gradient descent from one prediction to multiple predictions.

```python
def ele_mul(number, vector):
    return [number * x for x in vector]


def neural_network(input, weights):
    return ele_mul(input, weights)


input = 0.65
weights = [0.3, 0.2, 0.9]
true = [0.1, 1.0, 0.1]
alpha = 0.1

pred = neural_network(input, weights)
delta = [pred[i] - true[i] for i in range(len(true))]
weight_deltas = ele_mul(input, delta)

for i in range(len(weights)):
    weights[i] -= alpha * weight_deltas[i]

print("Predictions:", pred)
print("Updated weights:", weights)
```

## 6. Gradient Descent Learning with Multiple Inputs and Outputs

A network can combine multiple input features and generate multiple predictions. Each output uses its own set of weights for the input features.

The prediction process combines inputs and weights, while the learning process calculates errors for the outputs and updates the weights associated with them. This structure extends the same learning principles to a more flexible network.

```python
def vect_mat_mul(vect, matrix):
    assert len(vect) == len(matrix)
    output = []

    for row in matrix:
        output.append(w_sum(vect, row))

    return output


def neural_network(input, weights):
    return vect_mat_mul(input, weights)


weights = [
    [0.1, 0.1, -0.3],
    [0.1, 0.2, 0.0],
    [0.0, 1.3, 0.1]
]

input = [8.5, 0.65, 1.2]
pred = neural_network(input, weights)

print("Predictions:", pred)
```

## 7. What Do These Weights Learn?

Weights determine how strongly input features contribute to predictions. During training, their values change based on the prediction errors and the input values.

A weight's contribution depends on the data and its corresponding input. Different input magnitudes can cause different weights to receive updates of different sizes.

## 8. Visualizing Weight Values

Observing weight values during training helps explain how the network learns. Comparing weights across iterations shows which parameters change and how the learning process affects the model.

The changes in weights should be considered alongside prediction errors, because the purpose of updating weights is to improve predictions.

```python

def w_sum(a, b):
    assert len(a) == len(b)
    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output


def neural_network(input, weights):
    return w_sum(input, weights)


weights = [0.1, 0.2, -0.1]
input = [8.5, 0.65, 1.2]
true = 1
alpha = 0.01

for iteration in range(10):
    pred = neural_network(input, weights)
    delta = pred - true
    weight_deltas = [x * delta for x in input]

    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]

    print(f"Iteration {iteration + 1}: {weights}")
```

## 9. Visualizing Dot Products (Weighted Sums)

A dot product multiplies corresponding elements from two vectors and adds the results. In a neural network, this operation combines input values with their weights to produce a prediction.

The weighted sum is an important operation for networks with multiple inputs because it provides a way to combine several features into one output.

```python
input = [8.5, 0.65, 1.2]
weights = [0.1, 0.2, -0.1]

products = [input[i] * weights[i] for i in range(len(input))]
weighted_sum = sum(products)

print("Individual products:", products)
print("Dot product:", weighted_sum)
```

## 10. Summary

Gradient descent can be generalized from a single weight to networks with multiple weights. Each weight receives an update based on its corresponding input and the error signal.

The same learning approach can be extended to networks with multiple inputs and multiple outputs. Weighted sums combine input features, while repeated weight updates help the network reduce prediction error.
