## Deep Learning on Unseen Data: Introducing Federated Learning

This chapter introduces federated learning, a technique for training machine learning models across distributed data sources.

Instead of collecting all raw data in one central location, participating clients can train models locally and share model updates with a central server.

The chapter explores privacy challenges, federated learning, spam detection, secure aggregation, and homomorphic encryption.


## The Problem of Privacy in Deep Learning

Machine learning models often require large amounts of data. However, collecting data from different users in one central location can create privacy and security concerns.

Personal information may be sensitive, and organizations may have legal or practical restrictions on sharing it.

Federated learning explores a different approach: training models across distributed data sources while limiting the need to transfer raw data to a central server.



## Federated Learning

Federated learning allows multiple clients to contribute to training a shared model.

Each client trains the model using its local dataset. The clients then send model updates to a coordinating server, which combines the updates to improve the shared model.

The process can repeat over multiple rounds until the model has learned useful patterns from the participating clients.


```python

import numpy as np

print("NumPy berhasil diimpor!")
```

```python

# Simulate data held by three clients
client_data = {
    "Client 1": np.array([1, 2, 3, 4]),
    "Client 2": np.array([5, 6, 7, 8]),
    "Client 3": np.array([9, 10, 11, 12])
}

for client, data in client_data.items():
    print(client, "local data:", data)
```


## Local Training

In federated learning, each client performs training using its own local data.

The client calculates a local model update based on its training examples. The raw data can remain on the client instead of being transmitted directly to the central server.

Keeping data local can reduce certain privacy risks, although model updates may still reveal information and require appropriate safeguards.


```python

def local_update(weight, data, learning_rate=0.01):
    gradient = np.mean(weight - data)
    return weight - learning_rate * gradient


global_weight = np.array([0.5])

for client, data in client_data.items():
    updated_weight = local_update(global_weight, data)
    print(client, "updated weight:", updated_weight)
```


## Aggregating Client Updates

After local training, clients send model updates to a coordinating server.

The server combines the updates to produce a new global model. A common approach is weighted averaging, in which clients with more training examples contribute proportionally more to the result.

The aggregated model can then be sent back to clients for another training round.


```python

global_weight = np.array([0.5])
learning_rate = 0.01

client_updates = []

for data in client_data.values():
    updated = local_update(
        global_weight,
        data,
        learning_rate
    )
    client_updates.append(updated)

global_weight = np.mean(client_updates, axis=0)

print("Aggregated global weight:", global_weight)
```


## Learning to Detect Spam

Spam detection is a useful example of a machine learning task that can involve distributed user data.

Different clients may have different message patterns. Federated learning can allow clients to contribute to a shared spam-detection model without directly centralizing all their messages.

The goal is to improve the model using information learned across clients while reducing the need to share raw message content.


```python

messages = [
    "free prize claim now",
    "meeting scheduled tomorrow",
    "win money now",
    "project discussion today"
]

labels = np.array([1, 0, 1, 0])

vocabulary = sorted(
    set(word for message in messages for word in message.split())
)

word2index = {
    word: i for i, word in enumerate(vocabulary)
}


def vectorize(message):
    vector = np.zeros(len(vocabulary))

    for word in message.split():
        vector[word2index[word]] += 1

    return vector


X = np.array([vectorize(message) for message in messages])

print("Vocabulary:", vocabulary)
print("Feature matrix shape:", X.shape)
print("Labels:", labels)
```


## Local Model Training for Spam Detection

A spam-detection model can be trained separately on each client's labeled messages.

The clients may have different amounts of data and different distributions of spam and legitimate messages.

Federated training combines the resulting model updates to create a shared model. In a real implementation, the model architecture, loss function, local optimization, and aggregation protocol must be specified carefully.



## Understanding Model Updates

A model update describes how the model parameters change during local training.

The server can combine updates from participating clients without directly receiving their original examples.

However, model updates are not automatically private. Depending on the training setup, they can leak information about the data used to create them.



## Secure Aggregation

Secure aggregation is a cryptographic technique that helps a server obtain an aggregate of client updates without directly seeing each client's individual update.

The protocol is designed so that the server learns the combined result rather than the individual contributions, subject to the protocol's assumptions and security guarantees.

Secure aggregation can improve privacy in federated learning, but it does not eliminate every possible privacy risk.


```python

# Simulated model updates from three clients
updates = [
    np.array([0.2, 0.4, 0.6]),
    np.array([0.1, 0.3, 0.5]),
    np.array([0.4, 0.2, 0.3])
]

# Demonstration only: ordinary averaging is not
# a cryptographic secure-aggregation protocol.
aggregated_update = np.mean(updates, axis=0)

print("Aggregated update:", aggregated_update)
```


## Homomorphic Encryption

Homomorphic encryption is a cryptographic approach that allows certain computations to be performed directly on encrypted data.

Depending on the scheme, computations on ciphertexts can correspond to operations on the underlying plaintext values.

In a federated learning context, homomorphic encryption may help protect model updates during aggregation. It typically introduces computational and communication overhead.



## Privacy and Security Considerations

Federated learning reduces the need to centralize raw training data, but it should not be treated as a complete privacy guarantee.

Practical systems may combine federated learning with secure aggregation, encryption, access controls, and differential privacy.

The appropriate protections depend on the sensitivity of the data, the threat model, the number of participating clients, and the system's requirements.
