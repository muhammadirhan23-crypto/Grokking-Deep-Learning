## 1. Predict, Compare, and Learn

Neural networks use three main steps: predict, compare, and learn.

- Predict: Generate a prediction from input data.
- Compare: Measure the difference between the prediction and the expected result.
- Learn: Adjust the weights to reduce the error.

This chapter focuses on comparing predictions and learning from errors.

## 2. Compare

The compare step measures how far a prediction is from the expected result.

The error measurement indicates whether the prediction is accurate or inaccurate. In this chapter, mean squared error is used to measure prediction error.

## 3. Learn

The learn step determines how the weights should change to reduce prediction error.

The network evaluates the effect of its weights and adjusts them to improve future predictions. Gradient descent is introduced as a method for determining these adjustments.

## 4. Compare: Does Your Network Make Good Predictions?

Prediction error can be calculated by subtracting the expected result from the prediction and squaring the difference.

The squared error is always nonnegative and gives larger errors greater influence.

```python
knob_weight = 0.5
input = 0.5
goal_pred = 0.8

pred = input * knob_weight
error = (pred - goal_pred) ** 2

print(error)
```

## 5. Why Measure Error?

Error measures how far a prediction is from the expected result.

The book uses mean squared error to ensure that errors are nonnegative and that larger errors receive greater emphasis.

Measuring error helps evaluate prediction accuracy and provides information needed for the learning process.

## 6. What's the Simplest Form of Neural Learning?

The simplest form of neural learning adjusts a weight to reduce error.

The network makes a prediction, measures the error, and changes the weight. This process is repeated until the prediction becomes more accurate.

## 7. Hot and Cold Learning

Hot and cold learning tests different weight values to determine which direction reduces error.

The weight is adjusted slightly upward or downward. The network compares the resulting errors and chooses the direction that improves the prediction.

## 8. Characteristics of Hot and Cold Learning

Hot and cold learning is straightforward but requires testing multiple weight values.

Its main idea is to change a weight, evaluate the resulting error, and repeat the process. This approach helps explain the basic idea of learning before introducing more efficient methods.

## 9. Calculating Both Direction and Amount from Error

Learning requires determining both the direction and the amount by which a weight should change.

The prediction error provides information about the size of the mistake, while the input and weight relationship helps determine how the weight should be adjusted.

## 10. One Iteration of Gradient Descent

One iteration of gradient descent makes a prediction, calculates its error, determines a weight adjustment, and updates the weight.

The update uses the input, prediction error, and learning rate to move the weight toward a value that reduces error.

```python
weight = 0.1
alpha = 0.01

def neural_network(input, weight):
    prediction = input * weight
    return prediction

number_of_toes = [8.5]
win_or_lose_binary = [1]

input = number_of_toes[0]
goal_pred = win_or_lose_binary[0]

pred = neural_network(input, weight)
error = (pred - goal_pred) ** 2
delta = pred - goal_pred

weight_delta = delta * input
weight -= alpha * weight_delta

print("Prediction:", pred)
print("Error:", error)
print("Updated weight:", weight)
```

## 11. Learning Is Just Reducing Error

The objective of neural learning is to reduce prediction error.

By repeatedly updating the weights, the network can produce predictions that are closer to the expected results. The learning process continues as long as the updates improve the prediction.

## 12. Let's Watch Several Steps of Learning

Repeated learning iterations allow the network to gradually improve its predictions.

At each iteration, the network calculates the prediction error and updates its weight. As learning progresses, the error should decrease when the updates are appropriate.

## 13. Why Does This Work? What Is weight_delta, Really?

The weight delta represents the proposed change to a weight.

It is calculated from the prediction error and the input. Its value determines the direction and magnitude of the weight update before the learning rate is applied.

## 15. A Box with Rods Poking Out of It

The book uses an analogy to explain weights and their relationship to predictions.

## 16. Derivatives: Take Two

Derivatives describe how a function changes when its input changes.

In neural learning, derivatives help determine how changing a weight affects prediction error. This information can be used to select a direction for reducing the error.

## 17. What You Really Need to Know

The important idea is that derivatives provide information about how a small change in a weight affects error.

A neural network can use this information to update weights in a direction that reduces the error.

## 18. What You Don't Really Need to Know

The chapter distinguishes between understanding the practical role of derivatives and mastering every mathematical detail.

For learning neural networks, the essential starting point is understanding how derivatives guide weight updates toward lower error.

## 19. How to Use a Derivative to Learn

A derivative can indicate how error changes as a weight changes.

Gradient descent uses this information to adjust the weight in the direction that reduces error. The learning rate controls the size of the update.

## 20. Look Familiar?

The learning process connects the earlier ideas of prediction, error measurement, and weight adjustment.

Gradient descent formalizes the process of using error-related information to improve predictions.

## 21. Breaking Gradient Descent

Gradient descent can fail to improve predictions when weight updates are too large.

An unsuitable update can move the weight past a useful value and increase the error instead of reducing it.

## 22. Visualizing the Overcorrections

Overcorrection occurs when a weight update moves too far past a value that would reduce error.

Repeated large updates can cause the network to move back and forth instead of converging toward a better prediction.

## 23. Divergence

Divergence occurs when the updates move the model away from a useful solution rather than toward it.

Large or poorly controlled weight changes can cause the error to grow instead of decrease.

## 24. Introducing alpha

Alpha represents the learning rate used to control the size of weight updates.

A smaller learning rate produces smaller changes, while a larger learning rate produces larger changes. Choosing an appropriate value helps prevent overcorrection.

```python
weight = 0.5
input = 0.5
goal_pred = 0.8
alpha = 0.01

pred = input * weight
error = (pred - goal_pred) ** 2
delta = pred - goal_pred

weight_delta = delta * input
weight -= alpha * weight_delta

print("Prediction:", pred)
print("Error:", error)
print("Updated weight:", weight)
```

## 25. Alpha in Code

The learning rate is implemented by multiplying the proposed weight update by alpha.

This controls how much the weight changes in each learning iteration. An appropriate learning rate helps gradient descent reduce error without making excessively large updates.

## 26. Memorizing

A neural network can reduce its error by adjusting its weights to fit the examples it has seen.

However, memorizing training examples does not necessarily mean the network will make accurate predictions on new data. This distinction becomes important when studying more advanced neural networks.
